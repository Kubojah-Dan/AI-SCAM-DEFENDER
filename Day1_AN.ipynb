```python
import pandas as pd
import numpy as np

df = pd.read_csv('student_performance.csv')


# Dataset Overview
print("Dataset Overview:")
print("Total Rows: ", df.shape[0])
print("Total Columns: ", df.shape[1])
print("Total Departments: ", df['department'].nunique())
```

```python
# Score Analysis
print("-----------------------")
high_socre = df['math_score'].max()
low_score = df['math_score'].min()
mean_score = df['math_score'].mean().round(2)

avg_pg = df['programming_score'].mean().round(2)
avg_att = df['attendance_percentage'].mean().round(2)

print("Highest Math Score: ", high_socre)
print("Lowest Math Score: ", low_score)
print("Average Math Score: ", mean_score)
print("Average Programming Score: ", avg_pg)
print(f"Average Attendance Percentage: {avg_att}%")
```

```python
# -------Gender-wise Analysis-----------
print("*" * 40)
print("\nGENDER-WISE AVERAGE SCORES")
df['total_score'] = df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']
gender_scores = df.groupby('gender')['total_score'].mean().round(2)
print(gender_scores)
```

```python
#----------Top Performers-----------

print("\nTOP PERFORMERS")
print("*" * 40)
top5 = df.sort_values('total_score', ascending=False).head(5)
for index, row in top5.iterrows():
    print(f"  {row['name']:<20} | {row['department']:<18} | Score: {row['total_score']}")
```

```python
#----------Students who need Attention---------------
print("\nSTUDENTS WITH LOW ATTENDANCE (Below 75%)")
print("*" * 40)
low_att = df[df['attendance_percentage'] < 75]
for index, row in low_att.iterrows():
    print(f"  {row['name']:<20} | {row['department']:<18} | Score: {row['total_score']}")
```
